```python
import sqlite3 as sql
```

```python
conn = sql.connect('chat_history.db')
```

```python
# list all tables
conn.execute("SELECT name FROM sqlite_master WHERE type='table';").fetchall()
```

```python
# list all columns in table 'messages'
conn.execute("PRAGMA table_info(messages);").fetchall()
```

```python
conn.execute("SELECT * FROM messages ORDER BY created_at DESC LIMIT 5;").fetchall()
```
